### Project Overview

Breaking into the corporate landscape as a fresher spans multiple avenues—from open job boards (LinkedIn, Naukri, Indeed) and niche startup portals to employee referrals and company drives.


This project conducts an in-depth Exploratory Data Analysis (EDA) on a dataset of **5,000 fresher job applications across 32 recruitment dimensions** time period of 2021 to 2024. Using Python, we evaluate what drives shortlisting and final hiring decisions, benchmark entry-level salary packages across roles and domains, and examine how hiring pipelines behave in today's entry-level job market.

### Key Analytical Objectives

* **Skill vs. Pedigree:** Does degree branch and CGPA dominate hiring decisions, or do hands-on projects, certifications, and internship history play a decisive role?
* **Recruitment Channel Performance:** Compare application response times and conversion rates across hiring platforms (LinkedIn, Naukri, Indeed, Internshala, AngelList, and Referrals).
* **The Hiring Funnel:** Map out drop-off and conversion rates across stages—from initial screening to online assessments, technical rounds, and final offers.
* **Compensation Benchmarking:** Identify high-paying sectors, in-demand fresher job roles, and regional salary variations across major employment hubs.
* **Network & Optimization Factor:** Measure whether LinkedIn activity, profile completeness, and referral usage correlate with faster response times and interview call-backs.

**Data Cleaning**

```python
# converting application_date to date_time
df['application_date'] = pd.to_datetime(df['application_date'])
df['application_date']
```

Handling missing value in the columns Interview_rounds,offered_salary_inr,linkedin_premium

```python
#handling interview round column filling 0 and need to convert the dtype float to int
df['interview_rounds'] = df['interview_rounds'].fillna(0).astype(int)
df['interview_rounds']
```

```python
#handling linkdin_premium missing values with "Unknown"
df['linkedin_premium'] = df['linkedin_premium'].fillna("Unknown")
df['linkedin_premium']
```

```python
df
```

```python
# 3. Standardize binary Yes/No columns into booleans
binary_cols = ['gap_year', 'prior_internship', 'referral_applied']
for col in binary_cols:
    df[col] = df[col].str.strip().map({'Yes': True, 'No': False})
```

```python
df['prior_internship']
```

```python
df['branch']
```

```python
# whitespace trimming across text columns
text_cols = df.select_dtypes(include='object').columns
for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
# standardize branch naming redundancies
branch_map = {
    'Mechanical': 'Mechanical Engineering',
    'Civil': 'Civil Engineering',
    'Electrical': 'Electrical Engineering'}

df['branch'] = df['branch'].replace(branch_map)
```

```python
print("\n==================================================")
print(" SECTOR-WISE SALARY & OPPORTUNITY DYNAMICS")
print("==================================================")

# Group rows by industry sector to analyze hiring volume, compensation, and academic profile requirements
sector_agg = (
    df.groupby("sector")
    .agg(
        total_applications=(
            "candidate_id",
            "count",
        ),  # Count total candidates per sector
        avg_salary=("offered_salary_inr", "mean"),  # Average offered salary
        max_salary=("offered_salary_inr", "max"),  # Peak salary offered
        avg_cgpa=("cgpa", "mean"),  # Average CGPA demanded/reported
    )
    .reset_index()
    .sort_values(
        by="avg_salary", ascending=False
    )  # Rank sectors from highest to lowest average salary
)
print(sector_agg.to_string(index=False))
```

```python
print("\n==================================================")
print(" RECRUITMENT PLATFORM PERFORMANCE")
print("==================================================")

# Group by application source platform to see which channels process the most volume and highest conversion
platform_agg = (
    df.groupby("platform")
    .agg(
        total_applications=(
            "candidate_id","count"
        ),  # Total candidate volume per portal
        avg_response_time=(
            "response_time_days",
            "mean",
        ),  # Speed of response
        offer_rate=(
            "has_salary_offer",
            lambda x: (x.mean()* 100)
        ),  # Conversion percentage into offers
    )
    .reset_index()
    .sort_values(
        by="total_applications", ascending=False
    )  # Sort by most popular platform first
)
print(platform_agg.to_string(index=False))
```

5. Multi-Line Chart: Monthly Application Trends by Work Type


Why this chart: Line charts capture continuous changes and momentum over chronological intervals, making it easy to identify trend convergences and seasonal spikes.

```python
df['application_month'] = pd.to_datetime(df['application_date']).dt.to_period('M').astype(str)
trend_data = df.groupby(['application_month', 'work_type']).size().unstack().fillna(0)

plt.figure(figsize=(11, 5))
for col in trend_data.columns:
    plt.plot(trend_data.index, trend_data[col], marker='o', markersize=3, label=col)
plt.title('Monthly Application Trend by Work Type', fontsize=13, fontweight='bold')
plt.xlabel('Application Month')
plt.ylabel('Applications Count')
plt.xticks(rotation=45, ticks=range(0, len(trend_data.index), 4), labels=trend_data.index[::4])
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

8. Grouped Bar Chart: Average Off-Campus Response Time by Work Type & Referral


Why this chart: Grouped bar charts compare a metric across two categorical dimensions side-by-side, making sub-segment differences immediately apparent.

```python
grouped = df.groupby(['work_type', 'referral_applied'])['response_time_days'].mean().unstack()

fig, ax = plt.subplots(figsize=(8, 5))
x = range(len(grouped.index))
width = 0.35

ax.bar([i - width/2 for i in x], grouped[False], width=width, label='No Referral', color='#8da0cb')
ax.bar([i + width/2 for i in x], grouped[True], width=width, label='Referral Applied', color='#fc8d62')
ax.set_xticks(x)
ax.set_xticklabels(grouped.index)
ax.set_title('Average Response Time: Referral vs. Non-Referral', fontsize=13, fontweight='bold')
ax.set_ylabel('Mean Response Time (Days)')
ax.legend()
plt.tight_layout()
plt.show()
```

Dataset Insight: Employee referrals do not significantly shorten recruiter response times in this pipeline, averaging roughly 22–24 days across both cohorts.

20. Catplot (Point Plot): Offer Rate by Gap Year and Prior Internship
Why this chart: Point plots show changes in point estimates and confidence intervals across factorial conditions, making interaction effects easy to spot.

```python
g = sns.catplot(
    data=df,
    x='gap_year',
    y='has_salary_offer',
    hue='prior_internship',
    kind='point',
    markers=['o', 's'],
    linestyles=['-', '--'],
    palette='Dark2',
    height=4.5,
    aspect=1.3
)
g.set_axis_labels('Took Gap Year', 'Offer Conversion Rate')
g._legend.set_title('Prior Internship')
plt.title('Offer Conversion: Interaction of Gap Year & Internship', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Conclusion & Hiring Intelligence Summary

Based on the exploratory data analysis of 5,000 fresher recruitment records, final hiring success and compensation are shaped by practical experience and strategic application channels rather than academic scores alone:


*   **Practical Experience Decisively Outweighs Pedigree**: While academic indicators like CGPA and backlogs show near-zero correlation with response times and salary offers, practical experience acts as the primary differentiator. Candidates holding prior internships show higher offer conversion resilience and exhibit a distinct secondary density peak toward premium entry-level compensation packages exceeding ₹8 LPA.
*   **High-Conversion Channels Outperform Volume Giants**: Open job boards such as LinkedIn and Naukri drive more than 41% of total applications, but specialized and direct sourcing channels deliver superior hireability. Direct employee referrals (15.42%) and early-career portals like Internshala (14.43%) achieve the highest offer conversion rates while cutting through the steep recruitment funnel.

*   **LinkedIn Activity & Profile Optimizations Show Diminishing Direct Impact**: A complete profile does not speed up initial ATS screening, with response times uniformly dispersed across 1 to 45 days regardless of fill percentage. Similarly, basic referral tags and standard profiles show marginal difference in overall speed (22–24 days), though candidates investing in LinkedIn Premium capture a slight lift in offer conversion (14.22% vs. 12.34%).
*   **Selective Sectors Drive Peak Entry-Level Compensation**: The overall salary offer rate hovers around 12.44% across the pipeline. While base packages consistently benchmark between ₹4.5L and ₹5.2L across industries, maximum entry-level compensation peaks significantly in IT/Software (₹19.26L), Manufacturing (₹18.90L), Retail (₹18.67L), and FinTech (₹17.86L).

*  ** Interview Rigor Correlates Directly with High-Value Roles**: Over 70% of candidate outcomes or rejections conclude within 0 to 2 interview rounds. The number of interview stages shows a distinct positive correlation (0.33) with offered salary, confirming that extended technical and domain evaluations are concentrated on higher-tier fresher roles.
